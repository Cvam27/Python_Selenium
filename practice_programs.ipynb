```python
a = 8
l1 = []
for i in range(1,a):
    if a%i==0:
        l1.append(i)
print(l1)
if sum(l1) == a:
    print("Perfect Number")
else:
    print("Not perfect")
```

```python
l1 = [1,2,3,4,5]

def findOdds(numbs):
    odds = []
    for i in numbs:
        if i%2!=0:
            odds.append(i)
    return odds
        
print(findOdds(l1))
```
